### Criar DataFrame para cada Continente

```python
idh_africas_list = [(k,v) for k,v in idh_africa.items()]
idh_africas_df = pd.DataFrame(idh_africas_list, columns=['País', 'IDH'])

idh_europa_list = [(k,v) for k,v in idh_europa.items()]
idh_europa_df = pd.DataFrame(idh_europa_list, columns=['País', 'IDH'])

idh_americas_list = [(k,v) for k,v in idh_america.items()]
idh_americas_df = pd.DataFrame(idh_americas_list, columns=['País', 'IDH'])

idh_asia_list = [(k,v) for k,v in idh_asia.items()]
idh_asia_df = pd.DataFrame(idh_asia_list, columns=['País', 'IDH'])

idh_oceania_list = [(k,v) for k,v in idh_oceania.items()]
idh_oceania_df = pd.DataFrame(idh_oceania_list, columns=['País', 'IDH'])
```

 ### Concatene os DataFrames em um único DataFrame usando o método pd.concat()

```python
idh_total_df = pd.concat([idh_africas_df, idh_europa_df, idh_americas_df, idh_asia_df, idh_oceania_df])
```
